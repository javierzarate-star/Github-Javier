```python
import random
def lanzar_dado_y_contar_suma_par(num_lanzamientos):
  """
  Simula el lanzamiento de dos dados y cuenta cuantas veces la suma de los resultados es par.
  Parameters:
  num_lanzamientos:el numero de dados de lanzamientos a simular.

  Returns
  La cantidad de veces que la suma de los resultados es par.
  """
  suma_par = 0 #Iniciamos contador para la suma par
  for _ in range(num_lanzamientos): # Iteramos sobre el número de lanzamientos
    dado1 = random.randint(1,6) #Generamos resultado primer dado.
    dado2 = random.randint(1,6) # Generamos resultado segundo dado
    suma = dado1+dado2 # Calculamos la suma de los resultados
    if suma % 2 == 0: # Verificamos si la suma es par
      suma_par += 1 # Incrementamos el contador si la suma es par
  return suma_par # Devolvemos la cantidad de sumas pares


  # Simulamos 100 lanzamientos y contamos la catidad de veces que la suma es par
num_lanzamientos = 200
resultado = lanzar_dado_y_contar_suma_par(num_lanzamientos)
print(f"De {num_lanzamientos} lanzamientos, {resultado} veces la suma fue par")
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

# Create a DataFrame for better viewing of the complete dataset
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df_iris['target'] = iris.target
df_iris['target_names'] = df_iris['target'].apply(lambda x: iris.target_names[x])

print("Iris Dataset (Complete):")
display(df_iris)
```

```python
from google.colab import sheets
sheet = sheets.InteractiveSheet(df=df_iris)
```

```python
print("Estadísticas descriptivas para el DataFrame de Iris:")
display(df_iris.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a dark theme for eye-catching visualization
plt.style.use('dark_background')

# --- Bar Chart 1: Distribution of Iris Species ---
plt.figure(figsize=(9, 6))
sns.countplot(x='target_names', data=df_iris, palette='viridis', edgecolor='white')

plt.title('Distribution of Iris Species', color='white', fontsize=18, fontweight='bold')
plt.xlabel('Iris Species', color='white', fontsize=14)
plt.ylabel('Count', color='white', fontsize=14)
plt.xticks(rotation=45, ha='right', color='white', fontsize=12)
plt.yticks(color='white', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6, color='gray') # Add a subtle grid
plt.tight_layout()
plt.show()

# --- Bar Chart 2: Average Petal Length per Iris Species ---
plt.figure(figsize=(9, 6))
sns.barplot(x='target_names', y='petal length (cm)', data=df_iris, palette='plasma', estimator='mean', errorbar='sd', edgecolor='white')

plt.title('Average Petal Length per Iris Species', color='white', fontsize=18, fontweight='bold')
plt.xlabel('Iris Species', color='white', fontsize=14)
plt.ylabel('Average Petal Length (cm)', color='white', fontsize=14)
plt.xticks(rotation=45, ha='right', color='white', fontsize=12)
plt.yticks(color='white', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6, color='gray')
plt.tight_layout()
plt.show()
```
